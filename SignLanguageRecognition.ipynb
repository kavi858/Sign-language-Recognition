```python
import pandas as pd

data = {
    "finger_count": [0, 1, 2, 3, 4, 5],
    "hand_open": [0, 0, 0, 0, 1, 1],
    "sign": ["A", "D", "V", "W", "B", "5"]
}

df = pd.DataFrame(data)

print("Sample Sign Language Dataset:")
print(df)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Select input features
X = df[["finger_count", "hand_open"]]

# Select target labels
y = df["sign"]

# Create and train model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print("AI Model Trained Successfully!")

# Test with sample data
prediction = model.predict(
    pd.DataFrame({
        "finger_count": [2],
        "hand_open": [0]
    })
)

print("Predicted Sign:", prediction[0])
```

```python
test_data = pd.DataFrame({
    "finger_count": [0, 1, 2, 3, 4, 5],
    "hand_open": [0, 0, 0, 0, 1, 1]
})

test_data["Predicted Sign"] = model.predict(test_data)

display(test_data)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 8))

plot_tree(
    model,
    feature_names=["finger_count", "hand_open"],
    class_names=model.classes_,
    filled=True,
    rounded=True,
    fontsize=10
)

plt.title("Sign Language Recognition - Decision Tree")
plt.show()
```

```python
from sklearn.tree import export_text

tree_rules = export_text(
    model,
    feature_names=["finger_count", "hand_open"]
)

print(tree_rules)
```

```python
import pandas as pd

new_sign = pd.DataFrame({
    "finger_count": [3],
    "hand_open": [0]
})

prediction = model.predict(new_sign)

print("AI Predicted Sign:", prediction[0])
```

```python
from sklearn.metrics import accuracy_score

X = df[["finger_count", "hand_open"]]
y = df["sign"]

predictions = model.predict(X)

accuracy = accuracy_score(y, predictions)

print("Model Accuracy:", accuracy * 100, "%")
```
